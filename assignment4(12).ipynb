```python
def processorder(price, quantity, customertype="regular"):
    total = price * quantity
    if total >= 10000:
        if customertype == "premium":
            total = total - (total * 0.20)
        else:
            total = total - (total * 0.10)
    if total >= 5000:
        shipping = 0
    else:
        shipping = 250
    finalamount = total + shipping
    return finalamount
print(processorder(2000, 6, "regular"))
print(processorder(2000, 6, "premium"))
print(processorder(500, 5, "regular"))
```
